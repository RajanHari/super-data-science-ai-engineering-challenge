```python
import random
from openai import OpenAI

client = OpenAI()

def roll_dice():
    return random.randint(1,6)
```

```python
target = 6
max_iterations = 20
```

```python
history = ""
interation = 0

while interation <= max_iterations:
    # ...
    interation+=1

    instructions = f"""
    
    You are playing a dice game. You need to roll a {target} on the dice to win. You've made the following rolls so far:
    {history if history else "No rolls yet."}
    Bsaed on what you have rolled so far, decide your next action:

    Option A: If you have achived the target, respond with exactly: DONE: <message describing how you achieved the target>
    Option B: If you have not achieved the target, respond with exactly: ROLL THE DICE

    """

    response = client.chat.completions.create(
        model='gpt-4.1-mini',
        messages=[{"role": "user", "content" : instructions}]
    )

    result = response.choices[0].message.content

    # Process the result to see if the agent has decided whether to roll again or not
    if "DONE:" in result:
        print(result)
        break

    if "ROLL THE DICE" in result:
        roll = roll_dice()
        print(f"Rolled: {roll}\n")
        history += f"Roll {interation}: {roll}\n"


print(f"Game ended after {interation-1} iterations.")
```
