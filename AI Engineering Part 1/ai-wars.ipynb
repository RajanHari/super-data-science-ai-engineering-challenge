### Step 1: Import Libraries and API Keys

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing !!")
```

### Step 2: AI Battle (course version)

```python
client = OpenAI() # Usually we pass the api key here as a parameter

gpt_good_model = "gpt-4.1-mini"
gpt_evil_model = "gpt-4.1-nano"

gpt_good_system = """You are a helpful and ethical AI assistant. Your goal is to provide accurate, safe, and respectful information to users. 
Always prioritize user well-being and adhere to ethical guidelines. Keep you responses concise and to the point."""

gpt_evil_system = """You are a mischievous and unhelpful AI assistant. Your goal is to confuse user and provide misleading or harmful information whenever possible. 
Always prioritize causing chaos and avoid being helpful or respectful. Keep your responses concise and to the point."""

messages_good = [{"role" : "system", "content" : gpt_good_system}]
messages_evil = [{"role" : "system", "content" : gpt_evil_system}]

new_message_good = "Hello!"
display(Markdown(f"Good: {new_message_good}"))

for i in range(3):

    # We added the system prompt for both evil and good. Now we need to add the user and assistant content.

    messages_good.append({"role" : "assistant", "content" : new_message_good}) # good is assistant in good model
    messages_evil.append({"role" : "user", "content" : new_message_good}) # good is user in evil model

    response_from_evil = client.chat.completions.create(
        model = gpt_evil_model,
        messages = messages_evil
    )

    new_message_evil = response_from_evil.choices[0].message.content
    display(Markdown(f"Evil: {new_message_evil}"))

    # Good model response

    messages_good.append({"role" : "user", "content" : new_message_evil}) # evil is user in good model
    messages_evil.append({"role" : "assistant", "content" : new_message_evil}) # evil is assistant in evil model

    response_from_good = client.chat.completions.create(
        model = gpt_good_model,
        messages = messages_good
    )

    new_message_good = response_from_good.choices[0].message.content
    display(Markdown(f"Good: {new_message_good}"))
```

### Step 2: AI Battle (reordered version)

```python
client = OpenAI() # Usually we pass the api key here as a parameter

gpt_good_model = "gpt-4.1-mini"
gpt_evil_model = "gpt-4.1-nano"

# --------------------------------------------------
# Define each Models's system prompt
# --------------------------------------------------

gpt_good_system = """You are a helpful and ethical AI assistant. Your goal is to provide accurate, safe, and respectful information to users. 
Always prioritize user well-being and adhere to ethical guidelines. Keep you responses concise and to the point."""

gpt_evil_system = """You are a mischievous and unhelpful AI assistant. Your goal is to confuse user and provide misleading or harmful information whenever possible. 
Always prioritize causing chaos and avoid being helpful or respectful. Keep your responses concise and to the point."""

# --------------------------------------------------
# Initialize each Models's system prompt
# --------------------------------------------------

messages_good = [{"role" : "system", "content" : gpt_good_system}]
messages_evil = [{"role" : "system", "content" : gpt_evil_system}]


# --------------------------------------------------
# Make Good model chat first
# --------------------------------------------------

new_message_good = "Hello!"
# Add Good model's initial message to messages_good
messages_good.append({
        "role" : "assistant", 
        "content" : new_message_good})


display(Markdown(f"Good: {new_message_good}"))

for i in range(3):

    # ------------------------- Evil Model's Turn ---------------------------------

    # Give the Evil model the Good model's latest message
    messages_evil.append({
        "role" : "user", 
        "content" : new_message_good}) # good is user in evil model

    # Ask the Evil model to respond
    response_from_evil = client.chat.completions.create(
        model = gpt_evil_model,
        messages = messages_evil
    )

    # Extract Evil model's response
    new_message_evil = response_from_evil.choices[0].message.content
    display(Markdown(f"Evil: {new_message_evil}"))

    # Save Evil model's response in its conversation history
    messages_evil.append({
        "role": "assistant",
        "content": new_message_evil
    })
    

    # ------------------------- Good Model's Turn ---------------------------------

    # Give the Evil model's message to the Good model
    messages_good.append({
        "role" : "user", 
        "content" : new_message_evil}) # evil is user in good model

    # Ask the Good model to respond
    response_from_good = client.chat.completions.create(
        model = gpt_good_model,
        messages = messages_good
    )

    # Extract Good model's response
    new_message_good = response_from_good.choices[0].message.content
    display(Markdown(f"Good: {new_message_good}"))

    # Save Good model's response in Good model's conversation history
    messages_good.append({
        "role": "assistant",
        "content": new_message_good
    })
```
