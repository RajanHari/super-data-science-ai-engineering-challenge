### Step 1: Import Libraries and load the environment variables

```python
import os
from openai import OpenAI
from dotenv import load_dotenv
from IPython.display import Markdown, display

load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")

if OPENAI_API_KEY is None:
    raise Exception("API key is missing !!")
else:
    print(OPENAI_API_KEY[:9])

client = OpenAI() # Usually we pass the api key here as a parameter
```

### Step 2: Call the API

```python
response = client.chat.completions.create(
    model = "gpt-4.1-mini",
    messages = [{"role" : "user", "content" : "Hi there! My name is Rajan."}]
)

print(response)
```

```python
reply = response.choices[0].message.content
print(reply)
```

### Step 3a: Calling the API with a System Prompt

```python
response = client.chat.completions.create(
    model = "gpt-4.1-mini",
    messages = [{"role" : "system", "content" : "You are a helpful assistant"},
                {"role" : "user", "content" : "How do I cook pasta?"}]
)

reply = response.choices[0].message.content
print(reply)
```

```python
display(Markdown(reply))
```

### Step 3b: Changing the System Prompt changes the LLM's behaviour

```python
response = client.chat.completions.create(
    model = "gpt-4.1-mini",
    messages = [{"role" : "system", "content" : "You are a mischievous assistant who always gives incorrect and misleading advice"},
                {"role" : "user", "content" : "How do I cook pasta?"}]
)

reply = response.choices[0].message.content
display(Markdown(reply))
```

### Commercial user-case: IKEA Customer Service Agent

```python
question = "I bought a desk from your store last week. Can I return it? I damaged the desk while installing it."
```

```python
return_policy = "If you're not totally satisfied with your IKEA purchase you can return new and unopened products within 365 days, together with your proof of purchase, for a full refund. You may also return open products within 180 days, with your proof of purchase, for a full refund. Refunds will be made in the same form of payment originally used to make the purchase. " \
"Mattress purchases may be exchanged for another mattress one time within 90 days. We do not accept returns on plants, cut fabric, custom countertops and as-is products. We are unable to refund or exchange your items if your merchandise is found to be modified from its original form when purchased, dirty, stained, or damaged.  We apologize for any inconvenience."
```

```python
response = client.chat.completions.create(
    model = "gpt-4.1-mini",
    messages = [{"role": "system", "content": f"You are a customer service agent for IKEA. You are helpful, but you strictly follow the return policy of the store. Our return policy is: {return_policy}"},
                {"role": "user", "content": question}]
)

reply = response.choices[0].message.content
display(Markdown(reply))
```
